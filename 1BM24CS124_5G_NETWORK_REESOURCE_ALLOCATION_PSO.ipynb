```python
import numpy as np

NUM_USERS = 5
TOTAL_BANDWIDTH = 100
NUM_PARTICLES = 30
MAX_ITERATIONS = 100
W = 0.7
C1 = 1.5
C2 = 1.5
ALPHA = 0.7
BETA = 0.3

channel_gain = np.array([1.2, 0.9, 1.5, 1.0, 0.8])

def fitness(position):
    bandwidth = np.maximum(position, 0)

    if np.sum(bandwidth) > TOTAL_BANDWIDTH:
        bandwidth = (
            bandwidth / np.sum(bandwidth)
            * TOTAL_BANDWIDTH
        )

    throughput = np.sum(
        channel_gain * np.log2(1 + bandwidth)
    )

    utilization = (
        np.sum(bandwidth) / TOTAL_BANDWIDTH
    )

    max_throughput = np.sum(
        channel_gain
        * np.log2(1 + TOTAL_BANDWIDTH)
    )

    normalized_throughput = (
        throughput / max_throughput
    )

    score = (
        ALPHA * normalized_throughput
        + BETA * utilization
    )

    return score, throughput, utilization

positions = np.random.uniform(
    0, TOTAL_BANDWIDTH / NUM_USERS,
    (NUM_PARTICLES, NUM_USERS)
)

velocities = np.random.uniform(
    -2, 2, (NUM_PARTICLES, NUM_USERS)
)

pbest = positions.copy()
pbest_scores = np.array([
    fitness(p)[0] for p in positions
])

best_index = np.argmax(pbest_scores)
gbest = pbest[best_index].copy()
gbest_score = pbest_scores[best_index]

for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = np.random.rand(NUM_USERS)
        r2 = np.random.rand(NUM_USERS)







        velocities[i] = (
            W * velocities[i]
            + C1 * r1 * (pbest[i] - positions[i])
            + C2 * r2 * (gbest - positions[i])
        )

        positions[i] += velocities[i]

        positions[i] = np.maximum(positions[i], 0)

        total = np.sum(positions[i])

        if total > TOTAL_BANDWIDTH:
            positions[i] *= TOTAL_BANDWIDTH / total

        score, _, _ = fitness(positions[i])

        if score > pbest_scores[i]:
            pbest[i] = positions[i].copy()
            pbest_scores[i] = score

    best_index = np.argmax(pbest_scores)
    if pbest_scores[best_index] > gbest_score:
        gbest = pbest[best_index].copy()
        gbest_score = pbest_scores[best_index]

best_score, throughput, utilization = fitness(gbest)

print("5G Network Resource Allocation using PSO")
print("----------------------------------------")
for i, bandwidth in enumerate(gbest):
    print(
        f"User {i + 1}: {bandwidth:.2f} MHz"
    )
print("----------------------------------------")
print(f"Total Bandwidth: {TOTAL_BANDWIDTH} MHz")
print(f"Allocated Bandwidth: {np.sum(gbest):.2f} MHz")
print(f"Throughput Metric: {throughput:.4f}")
print(f"Resource Utilization: {utilization * 100:.2f}%")
print(f"Best Fitness Score: {best_score:.4f}")
```
