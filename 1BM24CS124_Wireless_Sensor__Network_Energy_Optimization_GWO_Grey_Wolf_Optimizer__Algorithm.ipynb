```python
import random
import math

# Objective function
def fitness(wolf, target_pdr):
    power, distance = wolf

    pdr = 1 - math.exp(-5 * power / (distance / 10))

    hops = math.ceil(100 / distance)
    energy = hops * (power + 0.05)

    # Penalty for low packet delivery rate
    if pdr < target_pdr:
        energy += 1000 * (target_pdr - pdr) ** 2

    return energy


# USER INPUT
print("WIRELESS SENSOR NETWORK - GWO OPTIMIZATION")

N = int(input("Enter number of wolves: "))
MaxIter = int(input("Enter maximum iterations: "))

min_power = float(input("Enter minimum transmission power: "))
max_power = float(input("Enter maximum transmission power: "))

min_distance = float(input("Enter minimum communication distance: "))
max_distance = float(input("Enter maximum communication distance: "))

target_pdr = float(input("Enter required PDR (in percentage): ")) / 100

# Validate input
if N < 3 or MaxIter < 1:
    raise ValueError("Number of wolves must be at least 3 and iterations at least 1.")

if not (0 < min_power < max_power):
    raise ValueError("Invalid transmission power limits.")

if not (0 < min_distance < max_distance):
    raise ValueError("Invalid distance limits.")

if not (0 < target_pdr <= 1):
    raise ValueError("PDR must be between 0 and 100.")

lower = [min_power, min_distance]
upper = [max_power, max_distance]

# Initialize wolf population
wolves = [
    [
        random.uniform(lower[0], upper[0]),
        random.uniform(lower[1], upper[1])
    ]
    for _ in range(N)
]

# GREY WOLF OPTIMIZATION
for t in range(MaxIter):

    wolves.sort(key=lambda w: fitness(w, target_pdr))

    alpha = wolves[0][:]
    beta = wolves[1][:]
    delta = wolves[2][:]

    a = 2 - 2 * t / MaxIter

    new_wolves = []

    for wolf in wolves:

        positions = []

        for leader in [alpha, beta, delta]:

            new_position = []

            for d in range(2):

                r1 = random.random()
                r2 = random.random()

                A = 2 * a * r1 - a
                C = 2 * r2

                D = abs(C * leader[d] - wolf[d])

                X = leader[d] - A * D

                new_position.append(X)

            positions.append(new_position)

        # Update position
        new_wolf = [
            sum(p[d] for p in positions) / 3
            for d in range(2)
        ]

        new_wolf[0] = max(lower[0], min(upper[0], new_wolf[0]))
        new_wolf[1] = max(lower[1], min(upper[1], new_wolf[1]))

        new_wolves.append(new_wolf)

    wolves = new_wolves

wolves.sort(key=lambda w: fitness(w, target_pdr))

best = wolves[0]

power = best[0]
distance = best[1]

pdr = 1 - math.exp(-5 * power / (distance / 10))

hops = math.ceil(100 / distance)
energy = hops * (power + 0.05)

print("\n------ OPTIMIZATION RESULT ------")

print("Optimal Transmission Power:", round(power, 3))
print("Optimal Communication Distance:", round(distance, 2))
print("Minimum Energy Consumption:", round(energy, 3))
print("Packet Delivery Rate:", round(pdr * 100, 2), "%")

print("Network Lifetime (relative):", round(100 / energy, 2))

if pdr <= target_pdr:
    print("Status: Reliable Data Transmission")
else:
    print("Status: Required PDR not achieved")
```
