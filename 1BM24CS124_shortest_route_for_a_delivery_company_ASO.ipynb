```python
import random
import math

cities = ["A", "B", "C", "D", "E"]
distance = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 17],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 15],
    [25, 17, 20, 15, 0]
]
num_ants = 10
iterations = 50
alpha = 1
beta = 2
rho = 0.5
Q = 100
n = len(cities)
pheromone = [[1 for j in range(n)] for i in range(n)]
best_route = None
best_distance = float('inf')

for iteration in range(iterations):
    for ant in range(num_ants):
        route = [0]
        unvisited = list(range(1, n))

        while unvisited:
            current = route[-1]
            probabilities = []

            for city in unvisited:
                tau = pheromone[current][city] ** alpha
                eta = (1 / distance[current][city]) ** beta
                probabilities.append(tau * eta)

            total = sum(probabilities)
            probabilities = [p / total for p in probabilities]

            next_city = random.choices(
                unvisited, weights=probabilities
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)

        route.append(0)

        total_distance = sum(
            distance[route[i]][route[i + 1]]
            for i in range(n)
        )
        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

        for i in range(n):
            a = route[i]
            b = route[i + 1]
            pheromone[a][b] += Q / total_distance
            pheromone[b][a] += Q / total_distance

    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

print("Best Route:")
print(" -> ".join(cities[i] for i in best_route))

print("Shortest Distance:", best_distance)

```
