```python
import random
import math

# Objective function
def fitness(nest):
    power, distance = nest

    pdr = 1 - math.exp(-4 * power / (distance / 10))
    hops = math.ceil(100 / distance)

    energy = hops * (power + 0.05)

    # Penalty if PDR is below 85%
    penalty = 0 if pdr >= 0.85 else 1000000

    return energy + penalty, energy, pdr


# Initialize parameters
random.seed(42)

n = 20
iterations = 100
pa = 0.25

# Generate random nests
nests = [
    [random.uniform(0.1, 1),
     random.uniform(2, 10)]
    for _ in range(n)
]

# Cuckoo Search algorithm
for iteration in range(iterations):

    # Generate new solutions using Levy flight
    for i in range(n):

        step = random.gauss(0, 1) / (
            abs(random.gauss(0, 1)) ** (1 / 1.5) + 1e-9
        )

        new_nest = [
            max(0.1, min(1, nests[i][0] +
                0.08 * step * random.gauss(0, 1))),

            max(2, min(10, nests[i][1] +
                step * random.gauss(0, 1)))
        ]

        j = random.randrange(n)

        if fitness(new_nest)[0] < fitness(nests[j])[0]:
            nests[j] = new_nest

    # Sort nests based on fitness
    nests.sort(key=lambda x: fitness(x)[0])

    # Abandon worst nests
    for i in range(n - int(pa * n), n):
        nests[i] = [
            random.uniform(0.1, 1),
            random.uniform(2, 10)
        ]

    nests.sort(key=lambda x: fitness(x)[0])

# Find best solution
best = nests[0]

score, energy, pdr = fitness(best)

print("Wireless Sensor Network Energy Optimization")
print("--------------------------------------------")
print("Optimal Transmission Power:", round(best[0], 3))
print("Optimal Communication Distance:", round(best[1], 2))
print("Minimum Energy Consumption:", round(energy, 3))
print("Packet Delivery Rate:", round(pdr * 100, 2), "%")
print("Network Lifetime (relative):", round(100 / energy, 2))

```
