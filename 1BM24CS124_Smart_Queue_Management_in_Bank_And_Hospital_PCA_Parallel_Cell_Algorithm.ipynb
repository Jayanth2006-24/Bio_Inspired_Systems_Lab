```python
import random

def fitness(cells, arrival, service):
    total_wait = 0

    for queue in cells:
        queue.sort(key=lambda x: arrival[x])
        current_time = 0

        for customer in queue:
            start = max(current_time, arrival[customer])
            wait = start - arrival[customer]
            total_wait += wait
            current_time = start + service[customer]

    return total_wait / len(arrival)

print("PCA - SMART QUEUE MANAGEMENT")

n = int(input("Enter number of customers: "))
c = int(input("Enter number of service counters: "))
iterations = int(input("Enter number of iterations: "))

if n <= 0 or c <= 0 or iterations <= 0:
    raise ValueError("All values must be positive.")

arrival = []
service = []

for i in range(n):
    print("\nCustomer", i + 1)
    a = float(input("Enter arrival time: "))
    s = float(input("Enter service time: "))

    if a < 0 or s <= 0:
        raise ValueError("Arrival must be non-negative and service time positive.")

    arrival.append(a)
    service.append(s)

cells = [[] for _ in range(c)]

for customer in range(n):
    counter = random.randrange(c)
    cells[counter].append(customer)

for iteration in range(iterations):

    improved = False
    best_wait = fitness(cells, arrival, service)

    for i in range(c):

        neighbors = []

        if i > 0:
            neighbors.append(i - 1)

        if i < c - 1:
            neighbors.append(i + 1)

        for j in neighbors:

            for customer in cells[i][:]:

                cells[i].remove(customer)
                cells[j].append(customer)

                new_wait = fitness(
                    cells, arrival, service
                )

                if new_wait < best_wait:
                    best_wait = new_wait
                    improved = True
                else:
                    cells[j].remove(customer)
                    cells[i].append(customer)

    if not improved:
        break

print("\n----- OPTIMIZATION RESULT -----")

for i in range(c):
    print("Counter", i + 1, ":", [
        x + 1 for x in cells[i]
    ])

print("Average Waiting Time:",
      round(fitness(cells, arrival, service), 2))

print("Iterations Completed:", iteration + 1)

print("Status: Queue Optimization Completed")
```
